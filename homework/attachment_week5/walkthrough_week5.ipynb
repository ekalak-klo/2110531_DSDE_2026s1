Cluster edible mushrooms ด้วย 2 features แล้วแปลง centroid กลับ scale เดิม

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

FEATURES = ['cap-color-rate', 'stalk-color-above-ring-rate']
df = pd.read_csv('ModifiedEdibleMushroom.csv')
print(df.shape)
print(df['label'].value_counts(dropna=False))
```

## Q1 — edible only + 2 features + mean + StandardScaler

Expected: `(2104, 2)`

```python
edible = df.loc[df['label'] == 'e', FEATURES].copy()
print('rows after edible filter:', len(edible))

# fill missing with column mean (ก่อน scale)
edible = edible.fillna(edible.mean())

scaler = StandardScaler()
X = scaler.fit_transform(edible)
print('Q1 shape:', X.shape)
```

## Q2 — KMeans + max centroid (scaled)

`n_clusters=5, random_state=0, n_init='auto'`  
Expected: `[2.51 2.3 ]`

```python
km = KMeans(n_clusters=5, random_state=0, n_init='auto')
km.fit(X)
print('centroids (scaled):\n', km.cluster_centers_)

# max ต่อ feature ข้าม 5 clusters
q2 = np.round(km.cluster_centers_.max(axis=0), 2)
print('Q2:', q2)
```

## Q3 — inverse_transform → min centroid (original)

Expected: `[1.01 1.  ]`

```python
original = scaler.inverse_transform(km.cluster_centers_)
print('centroids (original):\n', original)

q3 = np.round(original.min(axis=0), 2)
print('Q3:', q3)
```
